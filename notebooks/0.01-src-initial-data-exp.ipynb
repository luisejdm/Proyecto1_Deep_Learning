```python
import tensorflow as tf
import matplotlib.pyplot as plt
```

```python
IMG_SIZE = (200, 300) # Suggested sizes by the dataset source
BATCH_SIZE = 64 # A batch size to make processing easier

dataset = tf.keras.utils.image_dataset_from_directory(
    "Motorbikes", # relative route for the 
    labels=None,
    image_size=IMG_SIZE,
    batch_size=BATCH_SIZE,
    shuffle=False
)
```

After importing the data, we find 798 different images we will visualize afterwards.

```python
for batch in dataset.take(1):
    images = batch

plt.figure(figsize=(10, 10))
for i in range(min(9, len(images))):
    plt.subplot(3, 3, i + 1)
    plt.imshow(images[i].numpy().astype("uint8"))
    plt.axis("off")

plt.show()
```

We have 4 variables in the specifications of our data. The first "None" relates to the arbitrary batch size we defined earlier, then 200 x 300 is the size in pixels for thge images and finally 3 refers to the 3 channels of the RGB component of the image.
